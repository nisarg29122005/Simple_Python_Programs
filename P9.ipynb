```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

# **Load Dataset**

```python
df = pd.read_csv('retail_sales_dataset.csv')

df.head()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.to_period('M')
```

```python
total_sales = df['Total Amount'].sum()
total_units = df['Quantity'].sum()

print("Total Sales:", total_sales)
print("Total Units Sold:", total_units)
```

# **Top 5 product by sales**

```python
top_products = df.groupby('Product Category')['Total Amount'].sum().nlargest(5)

top_products.plot(kind='bar')
plt.title('Top 5 Products by Sales')
plt.xlabel('Product Category')
plt.ylabel('Sales')
plt.show()
```

# **Sales by gender**

```python
region_sales = df.groupby('Gender')['Total Amount'].sum()

plt.figure()
region_sales.plot(kind='bar')
plt.title('Sales by Gender (Used instead of Region)')
plt.xlabel('Gender')
plt.ylabel('Sales')
plt.show()
```

# **Monthly sales trend**

```python
monthly_sales = df.groupby('Month')['Total Amount'].sum()

monthly_sales.plot(kind='line', marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
df['Month'] = df['Date'].dt.strftime('%Y-%m')
import plotly.express as px

monthly_sales = df.groupby('Month')['Total Amount'].sum().reset_index()

fig = px.line(monthly_sales, x='Month', y='Total Amount', title='Monthly Sales Trend')
fig.show()
```

# Study of Power BI Data Visualization Tool

## Abstract
This report offers a concise overview of Microsoft Power BI, a leading business intelligence tool. It covers its main features, components, and types of data visualizations. The report also highlights Power BI's advantages and limitations, provides a comparison with Google Colab, and illustrates its real-world applications. The aim is to emphasize the importance of data visualization in transforming raw data into actionable insights for informed decision-making.

## 1. Introduction to Power BI
In today's data-driven environment, understanding vast datasets is critical. Microsoft Power BI is a robust business intelligence (BI) tool that helps users analyze data and share insights. It converts raw data from various sources into interactive reports and dashboards, allowing users to connect, model, and visualize data easily without extensive technical expertise.

## 2. Key Features of Power BI
Power BI offers several powerful features:

*   **Data Connectivity:** Connects to diverse data sources (cloud, databases, files).
*   **Data Transformation:** Cleans and prepares data using Power Query.
*   **Data Modeling:** Builds relationships and calculates measures using DAX.
*   **Interactive Visualizations:** Provides dynamic charts, graphs, and maps.
*   **Collaboration:** Facilitates easy sharing and teamwork on reports.

## 3. Components of Power BI
Power BI is comprised of integrated components:

*   **Power BI Desktop:** A free Windows application for creating reports and data models.
*   **Power BI Service:** A cloud-based platform for sharing, collaborating, and managing reports.
*   **Power BI Mobile:** Apps for viewing reports and dashboards on mobile devices.

## 4. Types of Data Visualizations
Power BI supports various visualization types:

*   **Bar Chart:** Compares values across categories.
*   **Line Chart:** Shows trends over time.
*   **Pie Chart:** Displays proportions of a whole.
*   **Map:** Visualizes geographical data.
*   **KPI Cards:** Highlights key metrics and their status.

## 5. Advantages of Power BI

*   **User-Friendly:** Intuitive drag-and-drop interface.
*   **Cost-Effective:** Free Desktop version, competitive pricing for cloud service.
*   **Broad Connectivity:** Connects to numerous data sources.
*   **Interactive:** Creates dynamic and engaging dashboards.
*   **Regular Updates:** Frequent feature additions from Microsoft.

## 6. Limitations of Power BI

*   **Learning Curve:** Advanced features like DAX can be complex.
*   **Large Datasets:** Performance can be an issue with very large models.
*   **Customization:** Limited visual customization beyond default settings.
*   **No Offline Service:** Cloud service requires an internet connection.

## 7. Comparison between Power BI and Google Colab

| Feature           | Power BI                                  | Google Colab                                |
| :---------------- | :---------------------------------------- | :------------------------------------------ |
| **Primary Purpose** | Business Intelligence, Data Visualization | Python-based Data Analysis, Coding          |
| **User Interface**| Graphical User Interface (GUI)            | Code Editor for Python                      |
| **Ease of Use**   | Easy for visuals, harder for advanced modeling | Requires programming knowledge              |
| **Visualization** | Built-in interactive visuals              | Python libraries (Matplotlib, Seaborn)      |
| **Collaboration** | Built-in sharing                          | Shared notebooks via Google Drive           |
| **Primary Users** | Business analysts                         | Data scientists, developers                 |

## 8. Real-world Applications of Power BI
Power BI is used across industries for:

*   **Sales:** Analyzing performance and forecasting.
*   **Finance:** Monitoring budgets and financial reports.
*   **Healthcare:** Tracking patient outcomes and resources.
*   **Retail:** Understanding customer behavior and managing inventory.

## 9. Conclusion
Power BI significantly simplifies data analysis, making it accessible to a wide audience. Its features and integration capabilities solidify its role as a leading BI tool. While coding tools like Python in Google Colab offer flexibility for deep analysis, Power BI excels in presenting data visually, transforming raw data into actionable insights for better decision-making.
